A matrix $U$ is unitary if and only if its columns form an orthonormal basis of $\mathbb C^n$. Equivalently, $U^*U=I$, and

$$
\langle U\mathbf x,U\mathbf y\rangle=\langle\mathbf x,\mathbf y\rangle.
$$


```python
# Core imports used throughout Chapter 8
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
from IPython.display import display, Math, Markdown

sp.init_printing()
I = sp.I

def show_exact(expr, label=None):
    expr = sp.simplify(expr)
    if label:
        display(Math(rf"{label}={sp.latex(expr)}"))
    else:
        display(expr)
    return expr

def show_matrix(M, label=None):
    M = sp.Matrix(M)
    if label:
        display(Math(rf"{label}={sp.latex(M)}"))
    else:
        display(M)
    return M

def cinner(u, v):
    # Complex Euclidean inner product, linear in the first argument.
    u = sp.Matrix(u)
    v = sp.Matrix(v)
    return sp.simplify(sum(u[j] * sp.conjugate(v[j]) for j in range(len(u))))

def cnorm(u):
    return sp.sqrt(sp.simplify(cinner(u, u)))

def plot_complex_numbers(values, labels=None, title="Complex numbers in the complex plane"):
    values = [complex(sp.N(z)) for z in values]
    if labels is None:
        labels = [f"z{k+1}" for k in range(len(values))]
    fig, ax = plt.subplots(figsize=(7, 6))
    ax.axhline(0, linewidth=1)
    ax.axvline(0, linewidth=1)
    for z, label in zip(values, labels):
        ax.arrow(0, 0, z.real, z.imag, length_includes_head=True,
                 head_width=0.15, head_length=0.22)
        ax.scatter([z.real], [z.imag])
        ax.text(z.real + 0.12, z.imag + 0.12, label, fontsize=11)
    ax.set_xlabel("Real axis")
    ax.set_ylabel("Imaginary axis")
    ax.set_title(title)
    ax.grid(True, alpha=0.35)
    ax.set_aspect("equal", adjustable="box")
    plt.show()

def plot_roots(roots, title="Roots in the complex plane"):
    vals = [complex(sp.N(z)) for z in roots]
    fig, ax = plt.subplots(figsize=(6.5, 6.5))
    ax.axhline(0, linewidth=1)
    ax.axvline(0, linewidth=1)
    for k, z in enumerate(vals):
        ax.scatter(z.real, z.imag, s=65)
        ax.plot([0, z.real], [0, z.imag], linestyle="--", linewidth=1)
        ax.text(z.real + 0.08, z.imag + 0.08, f"k={k}")
    radius = abs(vals[0]) if vals else 1
    theta = np.linspace(0, 2*np.pi, 500)
    ax.plot(radius*np.cos(theta), radius*np.sin(theta), alpha=0.5)
    ax.set_aspect("equal", adjustable="box")
    ax.set_xlabel("Real axis")
    ax.set_ylabel("Imaginary axis")
    ax.set_title(title)
    ax.grid(True, alpha=0.35)
    plt.show()
```

# Section 8.1 - Complex Numbers

## Learning objectives

Students should be able to:

- write a complex number in standard form;
- identify its real and imaginary parts;
- represent a complex number as a point and vector in the complex plane;
- add, subtract, scale, and multiply complex numbers;
- solve quadratic equations with negative discriminant;
- perform matrix operations with complex entries;
- compute determinants of small complex matrices.

### Definition: complex number

> A complex number has the form $z=a+bi$, where $a,b\in\mathbb R$ and $i^2=-1$. The number $a$ is the real part, written $\operatorname{Re}(z)$, and $b$ is the imaginary coefficient, written $\operatorname{Im}(z)$.

## Why complex numbers are needed

The equation

$$
x^2+1=0
$$

has no real solution because every real square is nonnegative. Introducing a number $i$ satisfying $i^2=-1$ gives the two solutions $x=i$ and $x=-i$.

Every polynomial of positive degree has all of its roots in the complex numbers when roots are counted with multiplicity. This is the content of the Fundamental Theorem of Algebra. Complex numbers therefore provide an algebraically complete number system for polynomial equations.

The Fundamental Theorem of Algebra says that polynomials factor completely over $\mathbb C$ when roots are counted with multiplicity. This matters in linear algebra because a real matrix can have a characteristic polynomial with nonreal roots.


## Equality and uniqueness of standard form

If

$$
a+bi=c+di,
$$

then

$$
(a-c)+(b-d)i=0.
$$

The only way a real multiple of $1$ plus a real multiple of $i$ can equal zero is for both coefficients to vanish. Therefore,

$$
a=c
\qquad\text{and}\qquad
b=d.
$$

This proves that the representation $a+bi$ is unique.

## The complex plane

The complex number $z=a+bi$ corresponds to:

- the point $(a,b)$;
- the vector from $(0,0)$ to $(a,b)$.

The horizontal axis is the real axis, and the vertical axis is the imaginary axis. This representation lets us interpret addition as vector addition.

```python
# A fictional travel-and-music visualization
z1 = 4 + 3*I
z2 = -2 - I
z3 = -3*I
z4 = 5
plot_complex_numbers(
    [z1, z2, z3, z4],
    labels=[r"$4+3i$", r"$-2-i$", r"$-3i$", r"$5$"],
    title="Four complex numbers represented as vectors"
)
```

## Addition, subtraction, and real scalar multiplication

Let $z=a+bi$ and $w=c+di$. Then

$$
z+w=(a+c)+(b+d)i,
$$

$$
z-w=(a-c)+(b-d)i,
$$

and for a real scalar $k$,

$$
kz=ka+kbi.
$$

These rules are exactly the componentwise rules for vectors in $\mathbb R^2$.

### Detailed example

Let

$$
z=4+3i,
\qquad
w=-2+5i.
$$

Then

$$
\begin{aligned}
z+w
&=(4+3i)+(-2+5i)\\
&=(4-2)+(3+5)i\\
&=2+8i,
\end{aligned}
$$

and

$$
\begin{aligned}
z-w
&=(4+3i)-(-2+5i)\\
&=(4+2)+(3-5)i\\
&=6-2i.
\end{aligned}
$$

## Multiplication of complex numbers

Use distributivity and replace $i^2$ by $-1$:

$$
\begin{aligned}
(a+bi)(c+di)
&=ac+adi+bci+bd i^2\\
&=ac-bd+(ad+bc)i.
\end{aligned}
$$

Thus,

$$
(a+bi)(c+di)=(ac-bd)+(ad+bc)i.
$$

### Complete multiplication example

For $z=4+3i$ and $w=-2+5i$,

$$
\begin{aligned}
zw
&=(4+3i)(-2+5i)\\
&=4(-2)+4(5i)+3i(-2)+3i(5i)\\
&=-8+20i-6i+15i^2\\
&=-8+14i-15\\
&=-23+14i.
\end{aligned}
$$

```python
z = 4 + 3*I
w = -2 + 5*I
print("z + w =", sp.expand(z+w))
print("z - w =", sp.expand(z-w))
print("z w   =", sp.expand(z*w))
```

## Quadratic equations and conjugate roots

Consider the fictional Spurs-score polynomial

$$
p(x)=x^2-6x+13.
$$

The discriminant is

$$
\Delta=b^2-4ac=(-6)^2-4(1)(13)=36-52=-16.
$$

Using $\sqrt{-16}=4i$,

$$
\begin{aligned}
x
&=\frac{6\pm\sqrt{-16}}{2}\\
&=\frac{6\pm4i}{2}\\
&=3\pm2i.
\end{aligned}
$$

### Theorem: conjugate-pair property for real polynomials

> If a polynomial has real coefficients and $z$ is a root, then $\overline z$ is also a root.

### Proof

Let

$$
p(x)=a_0+a_1x+\cdots+a_nx^n
$$

with every $a_j$ real. Suppose $p(z)=0$. Take the complex conjugate:

$$
\overline{p(z)}
=
\overline{a_0+a_1z+\cdots+a_nz^n}.
$$

Conjugation distributes over sums and products, and $\overline{a_j}=a_j$ because $a_j$ is real. Therefore,

$$
\overline{p(z)}
=
a_0+a_1\overline z+\cdots+a_n\overline z^{\,n}
=
p(\overline z).
$$

Because $p(z)=0$,

$$
p(\overline z)=\overline{p(z)}=\overline 0=0.
$$

Hence $\overline z$ is also a root.

## Complex matrices

A complex matrix is a matrix whose entries are complex numbers. Matrix addition, scalar multiplication, multiplication, transpose, and determinants are computed by the same algebraic rules used for real matrices, with complex arithmetic applied to the entries.

Before inverting a complex matrix, check $\det(A)\ne0$; the usual inverse theory remains valid over $\mathbb C$.


### Detailed matrix example

Let

$$
A=
\begin{bmatrix}
1+i&2\\
3-i&-i
\end{bmatrix},
\qquad
B=
\begin{bmatrix}
2&1-i\\
i&3
\end{bmatrix}.
$$

Their sum is

$$
A+B=
\begin{bmatrix}
3+i&3-i\\
3&3-i
\end{bmatrix}.
$$

The $(1,1)$ entry of $AB$ is

$$
(1+i)(2)+2i=2+2i+2i=2+4i.
$$

The $(1,2)$ entry is

$$
(1+i)(1-i)+2(3)=2+6=8.
$$

The remaining entries are computed in the same row-by-column manner.

```python
A = sp.Matrix([[1+I, 2], [3-I, -I]])
B = sp.Matrix([[2, 1-I], [I, 3]])
show_matrix(A+B, "A+B")
show_matrix(sp.simplify(A*B), "AB")
show_matrix(sp.simplify(B*A), "BA")
show_exact(A.det(), r"\det(A)")
```

The determinant is

$$
\begin{aligned}
\det(A)
&=(1+i)(-i)-2(3-i)\\
&=(1-i)-6+2i\\
&=-5+i.
\end{aligned}
$$

Because $\det(A)\neq0$, the matrix is invertible.

# Section 8.2 - Conjugates and Division of Complex Numbers

## Learning objectives

Students should be able to:

- compute and interpret complex conjugates;
- prove the principal algebraic properties of conjugation;
- calculate and interpret the modulus;
- divide complex numbers by rationalizing the denominator;
- prove product and quotient rules for modulus;
- find and verify the inverse of a complex matrix.

### Definition: complex conjugate

> For $z=a+bi$, the complex conjugate is $\overline z=a-bi$. Geometrically, conjugation reflects a point across the real axis.

## Fundamental conjugate identities

For all complex numbers $z$ and $w$,

$$
\overline{z+w}=\overline z+\overline w,
$$

$$
\overline{zw}=\overline z\,\overline w,
$$

$$
\overline{\overline z}=z,
$$

and, when $w\neq0$,

$$
\overline{\left(\frac zw\right)}
=
\frac{\overline z}{\overline w}.
$$

### Proof of the product identity

Let $z=a+bi$ and $w=c+di$. Then

$$
zw=(ac-bd)+(ad+bc)i.
$$

Therefore,

$$
\overline{zw}=(ac-bd)-(ad+bc)i.
$$

On the other hand,

$$
\begin{aligned}
\overline z\,\overline w
&=(a-bi)(c-di)\\
&=ac-adi-bci+bd i^2\\
&=(ac-bd)-(ad+bc)i.
\end{aligned}
$$

Thus,

$$
\overline{zw}=\overline z\,\overline w.
$$

### Definition: modulus

> For $z=a+bi$, the modulus is $|z|=\sqrt{a^2+b^2}$. It is the Euclidean distance from the origin to the point $(a,b)$.

The conjugate converts the two-dimensional magnitude calculation into a product:

$$
\begin{aligned}
z\overline z
&=(a+bi)(a-bi)\\
&=a^2-ab i+ab i-b^2i^2\\
&=a^2+b^2\\
&=|z|^2.
\end{aligned}
$$

Hence,

$$
|z|=\sqrt{z\overline z}.
$$

## Product rule for modulus

For complex numbers $z$ and $w$,

$$
|zw|=|z||w|.
$$

### Proof

Both sides are nonnegative, so it is enough to compare their squares:

$$
\begin{aligned}
|zw|^2
&=(zw)\overline{zw}\\
&=(zw)(\overline z\,\overline w)\\
&=(z\overline z)(w\overline w)\\
&=|z|^2|w|^2.
\end{aligned}
$$

Taking nonnegative square roots gives $|zw|=|z||w|$.

## Division by a complex number

To divide by $w=c+di\neq0$, multiply numerator and denominator by $\overline w=c-di$:

$$
\begin{aligned}
\frac{a+bi}{c+di}
&=
\frac{(a+bi)(c-di)}{(c+di)(c-di)}\\
&=
\frac{(a+bi)(c-di)}{c^2+d^2}.
\end{aligned}
$$

The denominator is now real and positive.

### Reciprocal formula

For $z=a+bi\ne0$,

$$
\frac1z=\frac{a-bi}{a^2+b^2}.
$$

Complex division is multiplication by this reciprocal.


### Detailed division example

$$
\begin{aligned}
\frac{5+7i}{2-i}
&=\frac{(5+7i)(2+i)}{(2-i)(2+i)}\\
&=\frac{10+5i+14i+7i^2}{4-i^2}\\
&=\frac{3+19i}{5}\\
&=\frac35+\frac{19}{5}i.
\end{aligned}
$$

```python
z = 5 + 7*I
w = 2 - I
show_exact(sp.simplify(z/w), r"\frac{5+7i}{2-i}")
print("Check by multiplying the quotient by the divisor:")
show_exact(sp.simplify((z/w)*w), "(z/w)w")
```

## Inverse of a complex matrix

The formula for a $2\times2$ inverse remains valid:

$$
\begin{bmatrix}
a&b\\c&d
\end{bmatrix}^{-1}
=
\frac{1}{ad-bc}
\begin{bmatrix}
d&-b\\-c&a
\end{bmatrix},
$$

provided $ad-bc\neq0$.

### Complete inverse example

Let

$$
A=
\begin{bmatrix}
1&i\\
i&1
\end{bmatrix}.
$$

First compute the determinant:

$$
\det(A)=1(1)-i(i)=1-i^2=2.
$$

Therefore,

$$
A^{-1}
=
\frac12
\begin{bmatrix}
1&-i\\
-i&1
\end{bmatrix}.
$$

Verification:

$$
\begin{aligned}
AA^{-1}
&=
\frac12
\begin{bmatrix}
1&i\\
i&1
\end{bmatrix}
\begin{bmatrix}
1&-i\\
-i&1
\end{bmatrix}\\
&=
\frac12
\begin{bmatrix}
1-i^2&-i+i\\
i-i&-i^2+1
\end{bmatrix}\\
&=
\frac12
\begin{bmatrix}
2&0\\0&2
\end{bmatrix}
=
I.
\end{aligned}
$$

```python
A = sp.Matrix([[1, I], [I, 1]])
Ainv = A.inv()
show_matrix(Ainv, "A^{-1}")
show_matrix(sp.simplify(A*Ainv), "AA^{-1}")
show_matrix(sp.simplify(Ainv*A), "A^{-1}A")
```

# Section 8.3 - Polar Form and De Moivre's Theorem

## Learning objectives

Students should be able to:

- convert between rectangular and polar forms;
- determine a principal argument using the correct quadrant;
- multiply and divide complex numbers in polar form;
- prove and apply De Moivre's theorem;
- derive and compute all $n$th roots of a nonzero complex number;
- interpret powers and roots geometrically.

### Argument versus principal argument

If $z=re^{i\theta}$, then every $\theta+2\pi k$ represents the same complex number. The **argument** is therefore a family of angles, while the **principal argument** chooses one representative, commonly from $(-\pi,\pi]$.


### Definition: polar form

> For $z=a+bi\neq0$, let $r=|z|$ and let $\theta$ be an argument of $z$. Then
> $z=r(\cos\theta+i\sin\theta)$. We also write $z=r\,\operatorname{cis}\theta$.

The rectangular and polar coordinates are related by

$$
a=r\cos\theta,
\qquad
b=r\sin\theta,
\qquad
r=\sqrt{a^2+b^2}.
$$

The argument is not unique because adding $2\pi k$ does not change sine or cosine. The principal argument satisfies

$$
-\pi<\operatorname{Arg}(z)\le\pi.
$$

## Quadrant analysis is essential

The equation $\tan\theta=b/a$ does not identify the quadrant by itself. A correct procedure is:

1. Plot or inspect the signs of $a$ and $b$.
2. Compute a reference angle.
3. Adjust the angle to the correct quadrant.
4. Choose the representative in $(-\pi,\pi]$ when the principal argument is requested.

### Detailed conversion example

Let

$$
z=-\sqrt3+i.
$$

Then

$$
r=\sqrt{(-\sqrt3)^2+1^2}=\sqrt4=2.
$$

The point lies in Quadrant II. The reference angle is $\pi/6$, so

$$
\theta=\pi-\frac{\pi}{6}=\frac{5\pi}{6}.
$$

Therefore,

$$
z=2\left(\cos\frac{5\pi}{6}+i\sin\frac{5\pi}{6}\right).
$$

### Convert back to rectangular form

Use

$$
x=r\cos\theta,\qquad y=r\sin\theta.
$$

For example,

$$
4\left(\cos\frac{\pi}{3}+i\sin\frac{\pi}{3}\right)
=
2+2\sqrt3\,i.
$$


```python
z = -sp.sqrt(3) + I
r = sp.sqrt(sp.re(z)**2 + sp.im(z)**2)
theta = sp.arg(z)
show_exact(r, "r")
show_exact(theta, r"\operatorname{Arg}(z)")
plot_complex_numbers([z], labels=[r"$-\sqrt3+i$"], title="Rectangular and polar interpretation")
```

## Multiplication in polar form

Let

$$
z_1=r_1(\cos\theta_1+i\sin\theta_1),
\qquad
z_2=r_2(\cos\theta_2+i\sin\theta_2).
$$

Then

$$
\begin{aligned}
z_1z_2
&=r_1r_2[
\cos\theta_1\cos\theta_2
-\sin\theta_1\sin\theta_2\\
&\qquad\qquad
+i(\sin\theta_1\cos\theta_2+\cos\theta_1\sin\theta_2)]\\
&=r_1r_2[
\cos(\theta_1+\theta_2)+i\sin(\theta_1+\theta_2)].
\end{aligned}
$$

Thus multiplication multiplies moduli and adds arguments.

Similarly,

$$
\frac{z_1}{z_2}
=
\frac{r_1}{r_2}
\left[
\cos(\theta_1-\theta_2)+i\sin(\theta_1-\theta_2)
\right],
\qquad z_2\neq0.
$$

Division divides moduli and subtracts arguments.

### Badminton-swing phasor example

Suppose two fictional swing components are

$$
z_1=2\,\operatorname{cis}\left(\frac{\pi}{6}\right),
\qquad
z_2=3\,\operatorname{cis}\left(-\frac{\pi}{4}\right).
$$

Then

$$
z_1z_2
=
6\,\operatorname{cis}\left(-\frac{\pi}{12}\right),
$$

and

$$
\frac{z_1}{z_2}
=
\frac23\,\operatorname{cis}\left(\frac{5\pi}{12}\right).
$$

### De Moivre's theorem

> For a positive integer $n$,
> $$
> [r(\cos\theta+i\sin\theta)]^n
> =
> r^n[\cos(n\theta)+i\sin(n\theta)].
> $$

## Proof by induction

For $n=1$, the formula is immediate.

Assume the formula holds for some positive integer $n$:

$$
[r\,\operatorname{cis}\theta]^n
=
r^n\operatorname{cis}(n\theta).
$$

Then

$$
\begin{aligned}
[r\,\operatorname{cis}\theta]^{n+1}
&=
[r\,\operatorname{cis}\theta]^n
[r\,\operatorname{cis}\theta]\\
&=
r^n\operatorname{cis}(n\theta)
\cdot
r\operatorname{cis}\theta\\
&=
r^{n+1}\operatorname{cis}((n+1)\theta).
\end{aligned}
$$

Therefore the statement holds for $n+1$, and mathematical induction proves it for every positive integer $n$.

### Power example

Because

$$
1+i=\sqrt2\,\operatorname{cis}\left(\frac{\pi}{4}\right),
$$

De Moivre's theorem gives

$$
\begin{aligned}
(1+i)^8
&=(\sqrt2)^8
\operatorname{cis}\left(8\cdot\frac{\pi}{4}\right)\\
&=16\operatorname{cis}(2\pi)\\
&=16.
\end{aligned}
$$

## Derivation of the $n$th-root formula

Suppose

$$
w=\rho\,\operatorname{cis}\phi
$$

satisfies

$$
w^n=z=r\,\operatorname{cis}\theta.
$$

By De Moivre's theorem,

$$
w^n=\rho^n\operatorname{cis}(n\phi).
$$

Equality requires

$$
\rho^n=r
\quad\Longrightarrow\quad
\rho=r^{1/n},
$$

and

$$
n\phi=\theta+2\pi k.
$$

Therefore,

$$
\phi=\frac{\theta+2\pi k}{n}.
$$

The distinct roots are

$$
w_k
=
r^{1/n}
\operatorname{cis}
\left(
\frac{\theta+2\pi k}{n}
\right),
\qquad
k=0,1,\ldots,n-1.
$$

## Why there are exactly $n$ distinct roots

If $k$ and $\ell$ are different integers between $0$ and $n-1$, the argument difference is

$$
\frac{2\pi(k-\ell)}{n}.
$$

This cannot be an integer multiple of $2\pi$ unless $k-\ell$ is a multiple of $n$. But $|k-\ell|<n$, so that occurs only when $k=\ell$. Thus the listed roots are distinct.

Any other integer $k$ differs from one of $0,\ldots,n-1$ by a multiple of $n$, which adds an integer multiple of $2\pi$ to the angle and produces no new root.

### Fourth roots of $16$

Write

$$
16=16\,\operatorname{cis}(0).
$$

The fourth roots have modulus $16^{1/4}=2$ and arguments

$$
\frac{0+2\pi k}{4}
=
\frac{k\pi}{2},
\qquad
k=0,1,2,3.
$$

Therefore the roots are

$$
2,\qquad 2i,\qquad -2,\qquad -2i.
$$

```python
roots = [2*sp.exp(I*sp.pi*k/2) for k in range(4)]
print([sp.simplify(r) for r in roots])
plot_roots(roots, title="The fourth roots of 16")
print("Verification:")
for r in roots:
    print(sp.simplify(r**4))
```

# Section 8.4 - Complex Vector Spaces and Inner Products

## Learning objectives

Students should be able to:

- perform vector operations with complex scalars;
- determine whether a set is a basis for $\mathbb C^n$;
- compute complex Euclidean inner products;
- explain why conjugation is required;
- calculate norms, distances, angles, and projections;
- apply complex Gram-Schmidt;
- verify the axioms of a complex inner product.

## Complex vector spaces

A complex vector space satisfies the same ten vector-space axioms as a real vector space, but its scalars are complex numbers.

A linear combination in a complex vector space has the form

$$
c_1\mathbf v_1+\cdots+c_k\mathbf v_k,
\qquad
c_j\in\mathbb C.
$$

This changes linear dependence: complex coefficients are allowed, so a set that is independent over $\mathbb R$ can become dependent over $\mathbb C$.

### Vector-operation example

Let

$$
\mathbf v=(1+2i,\;3-i),
\qquad
\mathbf u=(-2+i,\;4).
$$

Then

$$
\mathbf v+\mathbf u
=
(-1+3i,\;7).
$$

Also,

$$
\begin{aligned}
(2+i)\mathbf v
&=((2+i)(1+2i),\;(2+i)(3-i))\\
&=(5i,\;7+i).
\end{aligned}
$$

```python
v = sp.Matrix([1+2*I, 3-I])
u = sp.Matrix([-2+I, 4])
show_matrix(v+u, r"\mathbf v+\mathbf u")
show_matrix(sp.simplify((2+I)*v), r"(2+i)\mathbf v")
show_matrix(sp.simplify(3*v-(5-I)*u), r"3\mathbf v-(5-i)\mathbf u")
```

## Basis example in $\mathbb C^3$

Consider

$$
\mathbf v_1=(i,0,0),
\qquad
\mathbf v_2=(i,i,0),
\qquad
\mathbf v_3=(0,0,i).
$$

Suppose

$$
c_1\mathbf v_1+c_2\mathbf v_2+c_3\mathbf v_3=\mathbf0.
$$

Comparing components gives

$$
(c_1+c_2)i=0,
\qquad
c_2i=0,
\qquad
c_3i=0.
$$

Because $i\neq0$, the second and third equations imply $c_2=c_3=0$. The first then gives $c_1=0$. The set is linearly independent. Since $\dim_{\mathbb C}\mathbb C^3=3$, it is a basis.

> **Complex Euclidean inner product**
>
> For $\mathbf u,\mathbf v\in\mathbb C^n$, we use
> $$
> \langle\mathbf u,\mathbf v\rangle
> =
> \sum_{j=1}^n u_j\overline{v_j}.
> $$
> This convention is linear in the first argument and conjugate-linear in the second.

## Why the second vector must be conjugated

If we used the real dot-product formula without conjugation, a nonzero vector could have a zero or nonreal squared length. For example,

$$
(1,i)\cdot(1,i)=1+i^2=0.
$$

That would incorrectly assign length zero to the nonzero vector $(1,i)$.

With conjugation,

$$
\langle(1,i),(1,i)\rangle
=
1\overline1+i\overline i
=
1+i(-i)
=
2>0.
$$

Conjugation is therefore essential for positive definiteness.

## Complex inner-product axioms

For complex scalars $\alpha$ and vectors $\mathbf u,\mathbf v,\mathbf w$:

1. **Conjugate symmetry**
   $$
   \langle\mathbf u,\mathbf v\rangle
   =
   \overline{\langle\mathbf v,\mathbf u\rangle}.
   $$

2. **Additivity in the first argument**
   $$
   \langle\mathbf u+\mathbf v,\mathbf w\rangle
   =
   \langle\mathbf u,\mathbf w\rangle
   +
   \langle\mathbf v,\mathbf w\rangle.
   $$

3. **Homogeneity in the first argument**
   $$
   \langle\alpha\mathbf u,\mathbf v\rangle
   =
   \alpha\langle\mathbf u,\mathbf v\rangle.
   $$

4. **Positive definiteness**
   $$
   \langle\mathbf u,\mathbf u\rangle\ge0,
   $$
   with equality only for $\mathbf u=\mathbf0$.

### Proof of conjugate symmetry for the Euclidean inner product

$$
\begin{aligned}
\overline{\langle\mathbf v,\mathbf u\rangle}
&=
\overline{
\sum_{j=1}^n v_j\overline{u_j}
}\\
&=
\sum_{j=1}^n \overline{v_j}u_j\\
&=
\sum_{j=1}^n u_j\overline{v_j}\\
&=
\langle\mathbf u,\mathbf v\rangle.
\end{aligned}
$$

### Proof of positive definiteness

$$
\langle\mathbf u,\mathbf u\rangle
=
\sum_{j=1}^n u_j\overline{u_j}
=
\sum_{j=1}^n |u_j|^2.
$$

Every term is a nonnegative real number, so the sum is nonnegative. The sum equals zero only when every $|u_j|^2=0$, which occurs only when every $u_j=0$. Hence $\mathbf u=\mathbf0$.

## Detailed inner-product example

Let

$$
\mathbf u=(2+i,\;1-i,\;-1+2i),
$$

$$
\mathbf v=(1-i,\;2,\;3+i).
$$

Then

$$
\begin{aligned}
\langle\mathbf u,\mathbf v\rangle
&=(2+i)\overline{(1-i)}
 +(1-i)\overline2
 +(-1+2i)\overline{(3+i)}\\
&=(2+i)(1+i)+2(1-i)+(-1+2i)(3-i)\\
&=2+8i.
\end{aligned}
$$

```python
u = sp.Matrix([2+I, 1-I, -1+2*I])
v = sp.Matrix([1-I, 2, 3+I])
show_exact(cinner(u, v), r"\langle\mathbf u,\mathbf v\rangle")
show_exact(cinner(v, u), r"\langle\mathbf v,\mathbf u\rangle")
show_exact(sp.conjugate(cinner(v, u)), r"\overline{\langle\mathbf v,\mathbf u\rangle}")
show_exact(cnorm(u), r"\|\mathbf u\|")
show_exact(cnorm(u-v), r"d(\mathbf u,\mathbf v)")
```

## Norm and distance

The norm induced by the inner product is

$$
\|\mathbf u\|
=
\sqrt{\langle\mathbf u,\mathbf u\rangle}
=
\sqrt{\sum_{j=1}^n|u_j|^2}.
$$

The distance is

$$
d(\mathbf u,\mathbf v)=\|\mathbf u-\mathbf v\|.
$$

Two vectors are orthogonal when

$$
\langle\mathbf u,\mathbf v\rangle=0.
$$

## Cauchy-Schwarz inequality

For vectors in a complex inner product space,

$$
|\langle\mathbf u,\mathbf v\rangle|
\le
\|\mathbf u\|\,\|\mathbf v\|.
$$

### Detailed proof

If $\mathbf v=\mathbf0$, both sides are zero. Assume $\mathbf v\neq\mathbf0$ and define

$$
\alpha=\frac{\langle\mathbf u,\mathbf v\rangle}
{\langle\mathbf v,\mathbf v\rangle}.
$$

Positive definiteness gives

$$
0\le
\|\mathbf u-\alpha\mathbf v\|^2.
$$

Expand:

$$
\begin{aligned}
\|\mathbf u-\alpha\mathbf v\|^2
&=
\langle\mathbf u-\alpha\mathbf v,
\mathbf u-\alpha\mathbf v\rangle\\
&=
\langle\mathbf u,\mathbf u\rangle
-\overline\alpha\langle\mathbf u,\mathbf v\rangle
-\alpha\langle\mathbf v,\mathbf u\rangle
+|\alpha|^2\langle\mathbf v,\mathbf v\rangle.
\end{aligned}
$$

Substituting the chosen $\alpha$ and using conjugate symmetry gives

$$
0\le
\|\mathbf u\|^2
-
\frac{|\langle\mathbf u,\mathbf v\rangle|^2}
{\|\mathbf v\|^2}.
$$

Therefore,

$$
|\langle\mathbf u,\mathbf v\rangle|^2
\le
\|\mathbf u\|^2\|\mathbf v\|^2.
$$

Taking nonnegative square roots proves the result.

## Orthogonal projection

For $\mathbf v\neq0$, the projection of $\mathbf u$ onto the line spanned by $\mathbf v$ is

$$
\operatorname{proj}_{\mathbf v}\mathbf u
=
\frac{\langle\mathbf u,\mathbf v\rangle}
{\langle\mathbf v,\mathbf v\rangle}\mathbf v.
$$

The error vector is orthogonal to $\mathbf v$:

$$
\left\langle
\mathbf u-
\operatorname{proj}_{\mathbf v}\mathbf u,
\mathbf v
\right\rangle=0.
$$

### Projection verification

Let

$$
\mathbf u=(2+i,1),
\qquad
\mathbf v=(1,i).
$$

Then

$$
\langle\mathbf u,\mathbf v\rangle
=
(2+i)\overline1+1\overline i
=
2+i-i=2,
$$

and

$$
\langle\mathbf v,\mathbf v\rangle=2.
$$

Thus,

$$
\operatorname{proj}_{\mathbf v}\mathbf u
=
\mathbf v=(1,i).
$$

The residual is

$$
\mathbf u-\mathbf v=(1+i,1-i),
$$

and its inner product with $\mathbf v$ is zero.

```python
u = sp.Matrix([2+I, 1])
v = sp.Matrix([1, I])
projection = sp.simplify(cinner(u,v)/cinner(v,v))*v
residual = sp.simplify(u-projection)
show_matrix(projection, r"\operatorname{proj}_{\mathbf v}\mathbf u")
show_matrix(residual, r"\mathbf u-\operatorname{proj}_{\mathbf v}\mathbf u")
show_exact(cinner(residual, v), r"\langle\mathrm{residual},\mathbf v\rangle")
```

## Complex Gram-Schmidt enrichment

Start with independent vectors $\mathbf v_1,\mathbf v_2$. Define

$$
\mathbf w_1=\mathbf v_1,
$$

$$
\mathbf w_2
=
\mathbf v_2
-
\frac{\langle\mathbf v_2,\mathbf w_1\rangle}
{\langle\mathbf w_1,\mathbf w_1\rangle}
\mathbf w_1.
$$

Then $\mathbf w_1$ and $\mathbf w_2$ are orthogonal. Normalizing them gives an orthonormal basis.

For

$$
\mathbf v_1=(1,i),
\qquad
\mathbf v_2=(1,1),
$$

we obtain

$$
\langle\mathbf v_2,\mathbf v_1\rangle=1-i,
\qquad
\langle\mathbf v_1,\mathbf v_1\rangle=2.
$$

Therefore,

$$
\mathbf w_2
=
\begin{bmatrix}1\\1\end{bmatrix}
-
\frac{1-i}{2}
\begin{bmatrix}1\\i\end{bmatrix}
=
\frac{1+i}{2}
\begin{bmatrix}1\\-i\end{bmatrix}.
$$

```python
v1 = sp.Matrix([1, I])
v2 = sp.Matrix([1, 1])
w1 = v1
w2 = sp.simplify(v2 - cinner(v2,w1)/cinner(w1,w1)*w1)
u1 = sp.simplify(w1/cnorm(w1))
u2 = sp.simplify(w2/cnorm(w2))
show_matrix(w2, r"\mathbf w_2")
show_matrix(u1, r"\mathbf u_1")
show_matrix(u2, r"\mathbf u_2")
show_exact(cinner(u1,u2), r"\langle\mathbf u_1,\mathbf u_2\rangle")
show_exact(cinner(u1,u1), r"\|\mathbf u_1\|^2")
show_exact(cinner(u2,u2), r"\|\mathbf u_2\|^2")
```

# Section 8.5 - Unitary and Hermitian Matrices

## Learning objectives

Students should be able to:

- compute the conjugate transpose;
- prove and use conjugate-transpose identities;
- determine whether a matrix is unitary;
- explain norm and inner-product preservation by unitary matrices;
- recognize Hermitian matrices;
- prove that Hermitian eigenvalues are real;
- prove orthogonality of eigenvectors for distinct eigenvalues;
- unitarily diagonalize a Hermitian matrix.

### Definition: conjugate transpose

> For a complex matrix $A$, the conjugate transpose $A^*$ is obtained by transposing $A$ and conjugating every entry:
> $$
> A^*=\overline{A}^{\,T}.
> $$

### Entrywise description

If $A=[a_{ij}]$ is an $m\times n$ matrix, then $A^*$ is an $n\times m$ matrix with entries

$$
(A^*)_{ij}=\overline{a_{ji}}.
$$

For

$$
A=
\begin{bmatrix}
1+i&2-i&3\\
i&-1&4+2i
\end{bmatrix},
$$

$$
A^*
=
\begin{bmatrix}
1-i&-i\\
2+i&-1\\
3&4-2i
\end{bmatrix}.
$$

## Conjugate-transpose properties

For compatible complex matrices $A$ and $B$ and a complex scalar $c$,

$$
(A+B)^*=A^*+B^*,
$$

$$
(cA)^*=\overline c\,A^*,
$$

$$
(AB)^*=B^*A^*,
$$

$$
(A^*)^*=A.
$$

### Proof that $(AB)^*=B^*A^*$

The $(i,j)$ entry of $(AB)^*$ is

$$
\overline{(AB)_{ji}}
=
\overline{\sum_k a_{jk}b_{ki}}
=
\sum_k \overline{a_{jk}}\;\overline{b_{ki}}.
$$

The $(i,j)$ entry of $B^*A^*$ is

$$
\sum_k (B^*)_{ik}(A^*)_{kj}
=
\sum_k \overline{b_{ki}}\;\overline{a_{jk}}.
$$

Complex scalar multiplication is commutative, so these sums are equal entry by entry. Therefore,

$$
(AB)^*=B^*A^*.
$$

The reversal of order is the matrix analogue of the transpose rule.

```python
A = sp.Matrix([[1+I, 2-I, 3], [I, -1, 4+2*I]])
Astar = sp.conjugate(A).T
show_matrix(A, "A")
show_matrix(Astar, "A^*")
```

### Definition: unitary matrix

> A square complex matrix $U$ is unitary when
> $$
> U^*U=UU^*=I.
> $$
> Equivalently, $U^{-1}=U^*$.

## Characterization by orthonormal columns

Let the columns of $U$ be $\mathbf u_1,\ldots,\mathbf u_n$. The $(i,j)$ entry of $U^*U$ is

$$
(U^*U)_{ij}
=
\langle\mathbf u_j,\mathbf u_i\rangle
$$

under the linear-first convention. Therefore,

$$
U^*U=I
$$

if and only if

$$
\langle\mathbf u_i,\mathbf u_j\rangle
=
\begin{cases}
1,&i=j,\\
0,&i\neq j.
\end{cases}
$$

Thus $U$ is unitary if and only if its columns form an orthonormal basis of $\mathbb C^n$.

## A unitary audio-mixing model

Consider the fictional two-channel music mixer

$$
U=
\frac1{\sqrt2}
\begin{bmatrix}
1&i\\
i&1
\end{bmatrix}.
$$

Its conjugate transpose is

$$
U^*
=
\frac1{\sqrt2}
\begin{bmatrix}
1&-i\\
-i&1
\end{bmatrix}.
$$

Then

$$
\begin{aligned}
U^*U
&=
\frac12
\begin{bmatrix}
1&-i\\
-i&1
\end{bmatrix}
\begin{bmatrix}
1&i\\
i&1
\end{bmatrix}\\
&=
\frac12
\begin{bmatrix}
2&0\\0&2
\end{bmatrix}
=
I.
\end{aligned}
$$

Therefore $U$ is unitary.

```python
U = sp.Matrix([[1, I], [I, 1]])/sp.sqrt(2)
show_matrix(sp.simplify(sp.conjugate(U).T*U), "U^*U")
show_matrix(sp.simplify(U*sp.conjugate(U).T), "UU^*")
```

## Unitary matrices preserve inner products and norms

For a unitary matrix $U$,

$$
\langle U\mathbf x,U\mathbf y\rangle
=
\langle\mathbf x,\mathbf y\rangle.
$$

Indeed,

$$
\begin{aligned}
\langle U\mathbf x,U\mathbf y\rangle
&=
\mathbf x^{T}U^{T}\overline{U\mathbf y}\\
&=
\mathbf x^{T}U^{T}\overline U\,\overline{\mathbf y}\\
&=
\mathbf x^{T}\overline{\mathbf y}\\
&=
\langle\mathbf x,\mathbf y\rangle.
\end{aligned}
$$

In particular,

$$
\|U\mathbf x\|^2
=
\langle U\mathbf x,U\mathbf x\rangle
=
\langle\mathbf x,\mathbf x\rangle
=
\|\mathbf x\|^2.
$$

Thus unitary transformations preserve lengths, distances, and orthogonality. They are the complex analogue of orthogonal transformations.

### Definition: Hermitian matrix

> A square complex matrix $H$ is Hermitian when
> $$
> H^*=H.
> $$
> The diagonal entries must be real, and entries symmetric across the diagonal must be complex conjugates.

For example,

$$
H=
\begin{bmatrix}
2&i\\
-i&2
\end{bmatrix}
$$

is Hermitian because

$$
H^*
=
\begin{bmatrix}
2&i\\
-i&2
\end{bmatrix}
=
H.
$$

### Theorem: Hermitian eigenvalues are real

> Every eigenvalue of a Hermitian matrix is real.

## Detailed proof

Let

$$
H\mathbf v=\lambda\mathbf v
$$

with $\mathbf v\neq0$. Compute

$$
\langle H\mathbf v,\mathbf v\rangle
=
\lambda\langle\mathbf v,\mathbf v\rangle.
$$

Because $H$ is Hermitian,

$$
\langle H\mathbf v,\mathbf v\rangle
=
\langle\mathbf v,H\mathbf v\rangle.
$$

Using conjugate symmetry and $H\mathbf v=\lambda\mathbf v$,

$$
\begin{aligned}
\langle\mathbf v,H\mathbf v\rangle
&=
\langle\mathbf v,\lambda\mathbf v\rangle\\
&=
\overline\lambda\langle\mathbf v,\mathbf v\rangle.
\end{aligned}
$$

Therefore,

$$
\lambda\langle\mathbf v,\mathbf v\rangle
=
\overline\lambda\langle\mathbf v,\mathbf v\rangle.
$$

Since $\mathbf v\neq0$, positive definiteness gives $\langle\mathbf v,\mathbf v\rangle>0$. Dividing yields

$$
\lambda=\overline\lambda.
$$

A complex number equals its conjugate exactly when it is real. Hence $\lambda\in\mathbb R$.

### Theorem: distinct Hermitian eigenspaces are orthogonal

> Eigenvectors corresponding to distinct eigenvalues of a Hermitian matrix are orthogonal.

## Detailed proof

Let

$$
H\mathbf u=\lambda\mathbf u,
\qquad
H\mathbf v=\mu\mathbf v,
\qquad
\lambda\neq\mu.
$$

Because $H$ is Hermitian,

$$
\langle H\mathbf u,\mathbf v\rangle
=
\langle\mathbf u,H\mathbf v\rangle.
$$

The left side is

$$
\lambda\langle\mathbf u,\mathbf v\rangle.
$$

The right side is

$$
\overline\mu\langle\mathbf u,\mathbf v\rangle.
$$

Hermitian eigenvalues are real, so $\overline\mu=\mu$. Thus,

$$
(\lambda-\mu)\langle\mathbf u,\mathbf v\rangle=0.
$$

Because $\lambda\neq\mu$,

$$
\langle\mathbf u,\mathbf v\rangle=0.
$$

## Complete unitary diagonalization example

Let

$$
H=
\begin{bmatrix}
2&i\\
-i&2
\end{bmatrix}.
$$

The characteristic polynomial is

$$
\begin{aligned}
\det(H-\lambda I)
&=
\det
\begin{bmatrix}
2-\lambda&i\\
-i&2-\lambda
\end{bmatrix}\\
&=(2-\lambda)^2-i(-i)\\
&=(2-\lambda)^2-1\\
&=(\lambda-1)(\lambda-3).
\end{aligned}
$$

Thus the eigenvalues are $\lambda_1=1$ and $\lambda_2=3$.

### Eigenspace for $\lambda_1=1$

$$
H-I=
\begin{bmatrix}
1&i\\
-i&1
\end{bmatrix}.
$$

The first equation is

$$
x+iy=0,
$$

so $x=-iy$. Choose $y=1$:

$$
\mathbf v_1=
\begin{bmatrix}
-i\\1
\end{bmatrix}.
$$

Its norm is

$$
\|\mathbf v_1\|=\sqrt{|-i|^2+|1|^2}=\sqrt2.
$$

Therefore,

$$
\mathbf p_1
=
\frac1{\sqrt2}
\begin{bmatrix}
-i\\1
\end{bmatrix}.
$$

### Eigenspace for $\lambda_2=3$

$$
H-3I=
\begin{bmatrix}
-1&i\\
-i&-1
\end{bmatrix}.
$$

The first equation gives

$$
-x+iy=0,
$$

so $x=iy$. Choose $y=1$:

$$
\mathbf v_2=
\begin{bmatrix}
i\\1
\end{bmatrix}.
$$

After normalization,

$$
\mathbf p_2
=
\frac1{\sqrt2}
\begin{bmatrix}
i\\1
\end{bmatrix}.
$$

The normalized eigenvectors are orthogonal:

$$
\begin{aligned}
\langle\mathbf p_1,\mathbf p_2\rangle
&=
\frac12\left[
(-i)\overline i+1\overline1
\right]\\
&=
\frac12[(-i)(-i)+1]\\
&=0.
\end{aligned}
$$

Form

$$
P=
\frac1{\sqrt2}
\begin{bmatrix}
-i&i\\
1&1
\end{bmatrix}.
$$

Then $P$ is unitary and

$$
P^*HP=
\begin{bmatrix}
1&0\\0&3
\end{bmatrix}.
$$

```python
H = sp.Matrix([[2, I], [-I, 2]])
p1 = sp.Matrix([-I, 1])/sp.sqrt(2)
p2 = sp.Matrix([I, 1])/sp.sqrt(2)
P = sp.Matrix.hstack(p1, p2)
show_matrix(H, "H")
show_matrix(P, "P")
show_matrix(sp.simplify(sp.conjugate(P).T*P), "P^*P")
show_matrix(sp.simplify(sp.conjugate(P).T*H*P), "P^*HP")
```

## Spectral theorem for Hermitian matrices

Every Hermitian matrix $H$ has an orthonormal basis of eigenvectors. If these unit eigenvectors are placed in the columns of a unitary matrix $P$, then

$$
P^*HP=D
$$

is a real diagonal matrix.

Equivalently,

$$
H=PDP^*.
$$

This representation separates the transformation into:

1. a unitary change to eigenvector coordinates;
2. independent real scalings by the eigenvalues;
3. a unitary return to the original coordinates.

This is the complex spectral theorem and is the exact analogue of orthogonal diagonalization for real symmetric matrices.

### Proof of the complex spectral theorem

We prove the theorem by induction on the matrix size $n$.

**Base case:** For $n=1$, a Hermitian matrix is a real scalar $[\lambda]$, so the standard basis vector is an orthonormal eigenbasis.

**Inductive step:** Assume every $(n-1)\times(n-1)$ Hermitian matrix has an orthonormal eigenbasis. Let $H$ be an $n\times n$ Hermitian matrix.

Because the characteristic polynomial has a complex root, $H$ has an eigenvalue $\lambda_1$ and a nonzero eigenvector $\mathbf v_1$. The preceding theorem shows that $\lambda_1$ is real. Normalize $\mathbf v_1$ so that $\|\mathbf v_1\|=1$.

Consider the orthogonal complement

$$
W=\mathbf v_1^\perp
=
\{\mathbf w:\langle\mathbf w,\mathbf v_1\rangle=0\}.
$$

We first prove that $W$ is invariant under $H$. If $\mathbf w\in W$, then

$$
\begin{aligned}
\langle H\mathbf w,\mathbf v_1\rangle
&=\langle\mathbf w,H\mathbf v_1\rangle\\
&=\langle\mathbf w,\lambda_1\mathbf v_1\rangle\\
&=\overline{\lambda_1}\langle\mathbf w,\mathbf v_1\rangle\\
&=0.
\end{aligned}
$$

Thus $H\mathbf w\in W$.

The restriction of $H$ to the $(n-1)$-dimensional space $W$ is still Hermitian. By the induction hypothesis, $W$ has an orthonormal basis of eigenvectors

$$
\mathbf v_2,\ldots,\mathbf v_n.
$$

Because every vector in $W$ is orthogonal to $\mathbf v_1$, the set

$$
\{\mathbf v_1,\mathbf v_2,\ldots,\mathbf v_n\}
$$

is an orthonormal basis of $\mathbb C^n$ consisting entirely of eigenvectors of $H$.

Place these vectors in the columns of $P$. Then $P$ is unitary and

$$
P^*HP=D,
$$

where $D$ is the real diagonal matrix of eigenvalues. This completes the induction.

# Chapter 8 closing summary

The chapter begins by extending the scalar field from $\mathbb R$ to $\mathbb C$. Conjugation then becomes the central operation that makes division, modulus, inner products, and adjoints behave correctly.

The major structural progression is:

$$
\text{complex arithmetic}
\longrightarrow
\text{polar geometry}
\longrightarrow
\text{complex vector spaces}
\longrightarrow
\text{unitary and Hermitian matrix theory}.
$$

The real theory is recovered as a special case:

- real orthogonal matrices are unitary;
- real symmetric matrices are Hermitian;
- orthogonal diagonalization becomes unitary diagonalization.
